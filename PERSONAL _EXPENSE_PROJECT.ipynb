```python
import datetime
income=int(input('Enter your monthly income:   '))
print('---------------------')
print('Monthly Expense')
print('---------------------')
food=float(input('Enter your Food Expense:   '))
Rent=float(input('Enter your Rent Expense:   '))
Travel=float(input('Enter your Travel Expense:   '))
Shopping=float(input('Enter your Shopping Expense:   '))
Education=float(input('Enter your Education Expense:   '))
Other=float(input('Enter your other Expense:   '))
print('                   ')
total_expense=(food+Rent+Travel+Shopping+Education+Other)
print('total Expense:',total_expense)
print("                         ")
saving=income-total_expense
print('saving:',saving)
```
